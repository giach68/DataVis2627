```python
import pandas as pd
import matplotlib.pyplot as plt
iris = pd.read_csv("data/iris_csv.csv") 

fig, ax = plt.subplots()
colors = {'Iris-setosa':'red', 'Iris-versicolor':'green', 'Iris-virginica':'blue'}

ax.scatter(iris['petallength'], iris['petalwidth'],c=iris['class'].map(colors))

ax.grid(True)
plt.show()
```

```python
from sklearn.decomposition import PCA
X_reduced = PCA(n_components=2).fit_transform(iris.loc[:,["sepallength","sepalwidth","petallength","petalwidth"]])
fig, ax = plt.subplots()
colors = {'Iris-setosa':'red', 'Iris-versicolor':'green', 'Iris-virginica':'blue'}

ax.scatter(X_reduced[:,0],X_reduced[:,1],c=iris['class'].map(colors))
```

```python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt

features = iris[["sepallength", "sepalwidth", "petallength", "petalwidth"]]
X_reduced = TSNE(
    n_components=2,
    perplexity=30,
    init="pca",
    learning_rate="auto",
    random_state=42,
).fit_transform(features)

fig, ax = plt.subplots()
colors = {
    "Iris-setosa": "red",
    "Iris-versicolor": "green",
    "Iris-virginica": "blue",
}

ax.scatter(X_reduced[:, 0], X_reduced[:, 1],
           c=iris["class"].map(colors))
```
